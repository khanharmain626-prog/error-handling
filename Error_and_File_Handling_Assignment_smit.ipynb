### Q4 
Ek file `numbers.txt` banayein aur us mein 1 se 5 tak numbers line-by-line write karein.

```python
# Q4: Write numbers 1 to 5, one number per line
with open("numbers.txt", "w") as file:
    for number in range(1, 6):
        file.write(str(number) + "\n")

print("Numbers written successfully.")
```

### Q5
`numbers.txt` file ko read karein aur tamam numbers ka sum (total) calculate karke print karein.

```python
# Q5: Read numbers.txt and calculate the sum
with open("numbers.txt", "r") as file:
    numbers = [int(line.strip()) for line in file]

total = sum(numbers)
print("Numbers:", numbers)
print("Sum:", total)
```

### Q6 
Ek Python program likhein jo kisi file mein total lines count karke bataye.

```python
# Q6: Count the total number of lines in a file
with open("numbers.txt", "r") as file:
    line_count = sum(1 for line in file)

print("Total lines:", line_count)
```

### Q7 (Medium)
`with open()` statement ka istemal karte hue ek file `data.txt` read karein aur explain karein ki `with` ka kya faida hai.

```python
# Q7: Read data.txt using with open()
# Create a small example file so this cell can run independently.
with open("data.txt", "w") as file:
    file.write("Python file handling example.")

with open("data.txt", "r") as file:
    content = file.read()

print(content)
print("\nBenefit: The 'with' statement automatically closes the file,")
print("even if an error occurs inside the block.")
```

### Q8 (Medium)
Ek file `sample.txt` se read karke uski saari lines ko ek Python list mein convert karein (`readlines()` ka use karke).

### Q9 
Ek file `source.txt` ka data read karke `destination.txt` file mein copy karein.

```python
# Q9: Copy data from source.txt to destination.txt
# Create a source file for this example.
with open("source.txt", "w") as file:
    file.write("This is the source file.\nIts data will be copied.")

with open("source.txt", "r") as source_file:
    data = source_file.read()

with open("destination.txt", "w") as destination_file:
    destination_file.write(data)

print("File copied successfully.")
with open("destination.txt", "r") as file:
    print(file.read())
```

### Q10 
`os` module ka use karke check karein ki kya `test.txt` file exist karti hai ya nahi. Agar karti hai toh usay delete kar dein.

```python
# Q10: Check whether test.txt exists; if it does, delete it
import os

if os.path.exists("test.txt"):
    os.remove("test.txt")
    print("test.txt existed and has been deleted.")
else:
    print("test.txt does not exist.")
```

### Q11 
Aisa program likhein jo 10 ko 0 se divide kare aur `try-except` block ka use karke `ZeroDivisionError` ko handle kare.

```python
# Q11: Handle division by zero
try:
    result = 10 / 0
    print(result)
except ZeroDivisionError:
    print("Error: A number cannot be divided by zero.")
```

### Q12 (Easy)
User se number input lein. Agar user text enter kar de toh `ValueError` handle karke custom error message print karein.

```python
# Q12: Handle invalid text input with ValueError
try:
    number = int(input("Enter a number: "))
    print("You entered:", number)
except ValueError:
    print("Invalid input! Please enter a valid whole number.")
```

### Q13 (Easy)
Aisi file open karne ki koshish karein jo exist nahi karti (`non_existent_file.txt`). `FileNotFoundError` ko `try-except` se handle karein.

```python
# Q13: Handle a file that does not exist
try:
    with open("non_existent_file.txt", "r") as file:
        print(file.read())
except FileNotFoundError:
    print("Error: The requested file was not found.")
```

### Q14 
Ek program banayein jisme `finally` block ho, jo humesha execute ho chahe error aaye ya na aaye.

```python
# Q14: The finally block always executes
try:
    print("Inside the try block.")
    result = 10 / 2
    print("Result:", result)
except ZeroDivisionError:
    print("A division error occurred.")
finally:
    print("Finally block executed.")
```

### Q15 
Ek list `my_list = [10, 20, 30]` banayein. Invalid index (e.g., index 5) access karein aur `IndexError` catch karein.

```python
# Q15: Handle an invalid list index
my_list = [10, 20, 30]

try:
    print(my_list[5])
except IndexError:
    print("Error: The requested list index does not exist.")
```

### Q16 
Ek dictionary `student = {"name": "Ali"}` banayein. Non-existent key (`"age"`) access karein aur `KeyError` handle karein.

```python
# Q16: Handle a missing dictionary key
student = {"name": "Ali"}

try:
    print(student["age"])
except KeyError:
    print("Error: The key 'age' does not exist in the dictionary.")
```

### Q17
Ek `try-except` block likhein jo Multiple Exceptions catch kar sake (`ZeroDivisionError` aur `ValueError`).

```python
# Q17: Catch multiple exceptions
try:
    number = int(input("Enter a number: "))
    result = 10 / number
    print("Result:", result)
except (ZeroDivisionError, ValueError) as error:
    print("An error occurred:", error)
```

### Q18
`else` block ka istemal `try-except` ke sath karein jo sirf tab chale jab koi error na aaye.

```python
# Q18: The else block runs only when no exception occurs
try:
    number = int(input("Enter a number: "))
    result = 10 / number
except (ValueError, ZeroDivisionError) as error:
    print("Error:", error)
else:
    print("No error occurred. Result:", result)
```

### Q19 
`raise` keyword ka use karke age validation ka function banayein. Agar age 18 se kam ho toh custom `ValueError("Age must be 18 or older")` raise karein.

```python
# Q19: Validate age using raise
def validate_age(age):
    if age < 18:
        raise ValueError("Age must be 18 or older")
    return "Age is valid."

try:
    print(validate_age(20))
except ValueError as error:
    print("Error:", error)
```

### Q20 
File handling aur Exception handling ko combine karein: Ek file read karein, agar file na mile toh error catch karein, aur `finally` block mein file closure handle karein.

```python
# Q20: Combine file handling and exception handling
file = None

try:
    file = open("data.txt", "r")
    content = file.read()
    print(content)
except FileNotFoundError:
    print("Error: The file was not found.")
finally:
    if file is not None and not file.closed:
        file.close()
        print("File closed successfully.")
    else:
        print("No open file needed to be closed.")
```
